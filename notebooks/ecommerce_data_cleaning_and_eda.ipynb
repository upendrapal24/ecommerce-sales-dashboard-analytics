# E-commerce Sales Data Cleaning & Exploratory Data Analysis (EDA)
**Objective:** Clean, transform, and analyze the e-commerce sales dataset to uncover key revenue drivers, seller trends, product performance, and discount sensitivity.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3

# Graphics settings
sns.set_theme(style='darkgrid')
plt.rcParams['figure.figsize'] = (12, 6)
print('Libraries imported successfully.')
```

## 1. Load Raw Dataset

```python
df = pd.read_csv('../data/cleaned_ecommerce_data.csv')
print(f'Dataset Shape: {df.shape}')
df.head()
```

## 2. Dataset Overview & Data Quality Check

```python
df.info()
print('\nMissing Values:\n', df.isnull().sum())
print(f'\nDuplicates: {df.duplicated().sum()}')
```

## 3. Revenue & Category Analysis

```python
cat_summary = df.groupby('category').agg(
    total_revenue=('total_revenue', 'sum'),
    total_units=('units_sold', 'sum'),
    avg_price=('price', 'mean'),
    avg_discount=('discount_percent', 'mean'),
    avg_rating=('rating', 'mean')
).sort_values(by='total_revenue', ascending=False)

cat_summary['revenue_billions'] = cat_summary['total_revenue'] / 1e9
display(cat_summary)
```

## 4. Seller Performance Analysis

```python
seller_summary = df.groupby('seller').agg(
    total_revenue=('total_revenue', 'sum'),
    total_units=('units_sold', 'sum'),
    avg_seller_rating=('seller_rating', 'mean'),
    product_count=('product_id', 'count')
).sort_values(by='total_revenue', ascending=False)

seller_summary['revenue_billions'] = seller_summary['total_revenue'] / 1e9
display(seller_summary)
```

## 5. Discount Impact & Price Elasticity

```python
disc_summary = df.groupby('discount_tier').agg(
    product_count=('product_id', 'count'),
    total_revenue=('total_revenue', 'sum'),
    avg_units_sold=('units_sold', 'mean'),
    avg_rating=('rating', 'mean')
).reset_index()

display(disc_summary)
```
